# Probando NLTK en Python

Este es un Notebook para hacer pruebas sobre la utilización de NLTK en Python.

Vamos a realizar diferentes operaciones de procesamiento de texto con Lenguaje Natural, empleando dicha librería en Python.

```python
# Imporamos librerías necesarias
import nltk
from nltk.tokenize import sent_tokenize, word_tokenize
```

```python
# Generamos un texto largo de ejemplo
text = """El procesamiento de lenguaje natural (NLP) es un campo de la inteligencia artificial que se centra en la interacción entre las computadoras y el lenguaje humano. El objetivo del NLP es permitir que las computadoras comprendan, interpreten y generen lenguaje humano de manera que sea valiosa.
El NLP combina la lingüística computacional, que se basa en la gramática y el significado del lenguaje, con el aprendizaje automático, que permite a las computadoras aprender de los datos y mejorar su rendimiento con el tiempo.
El NLP se utiliza en una variedad de aplicaciones, como la traducción automática, el análisis de        sentimientos, la generación de texto y la respuesta a preguntas. Estas aplicaciones son cada vez más comunes en nuestra vida diaria, desde los asistentes virtuales hasta los motores de búsqueda.
El NLP enfrenta varios desafíos, como la ambigüedad del lenguaje, la variabilidad en la forma en que las personas se expresan y la necesidad de comprender el contexto. Sin embargo, los avances en el aprendizaje profundo y las redes neuronales han permitido que el NLP logre resultados impresionantes en tareas complejas.
El futuro del NLP es prometedor, con investigaciones en curso para mejorar la comprensión del lenguaje, la generación de texto y la interacción con los usuarios. A medida que las computadoras se vuelven más capaces de comprender el lenguaje humano, es probable que veamos un aumento en las aplicaciones de NLP en diversas industrias, desde la atención médica hasta el entretenimiento."""
```

```python
# Probando la función sent_tokenize para dividir el texto en oraciones
oraciones = sent_tokenize(text, language='spanish')

# Mostramos las oraciones obtenidas
for i, oracion in enumerate(oraciones):
    print(f"Oración {i+1}: {oracion}")
```

```python
# Vamos a tokenizar el texto en palabras
oraciones_en_palabras = []
# Recorremos cada oración y tokenizamos en palabras
for i, oracion in enumerate(oraciones):
    palabras = word_tokenize(oracion, language='spanish')
    oraciones_en_palabras.append(palabras)
    # Mostramos las palabras tokenizadas
    print(f"Oración {i+1} tokenizada en palabras: {palabras}")
```

## Graficos

```python
from nltk.text import Text

# Creamos un objeto Text de NLTK para trabajar con el texto tokenizado
texto = Text(word_tokenize(text, language='spanish'))
```

```python
# Mostramos una representación del texto
texto.dispersion_plot(['procesamiento', 'lenguaje', 'natural', 'inteligencia', 'artificial', 'NLP'])
```

```python
# Vamos a calcular las palabras más frecuentes en el texto
frecuencias = nltk.FreqDist(texto)
print("\nPalabras más frecuentes:")
for palabra, frecuencia in frecuencias.most_common(10):
    print(palabra, frecuencia)
```

```python
nltk.download('stopwords')
from nltk.corpus import stopwords
```

```python
# Vamos a eliminar las stopwords del texto
stop_words = set(stopwords.words('spanish'))

# Añadimos los signos de puntuación a las stopwords
import string
stop_words.update(string.punctuation)

# Filtramos las palabras del texto para eliminar las stopwords
texto_filtrado = [palabra for palabra in texto if palabra.lower() not in stop_words]

# Volvemos a calcular las frecuencias de las palabras filtradas
frecuencias_filtradas = nltk.FreqDist(texto_filtrado)

# Mostramos las palabras más frecuentes después de eliminar las stopwords
print("\nPalabras más frecuentes después de eliminar stopwords:")
for palabra, frecuencia in frecuencias_filtradas.most_common(10):
    print(palabra, frecuencia)
```

```python
# Graficamos la dispersión de las palabras filtradas
palabras_mas_comunes = [palabra for palabra, frecuencia in frecuencias_filtradas.most_common(10)]

texto.dispersion_plot(palabras_mas_comunes)
```
