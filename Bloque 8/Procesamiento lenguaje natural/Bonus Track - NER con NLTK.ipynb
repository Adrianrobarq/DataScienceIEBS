## 1. Reconocimiento de Entidades Nombradas con una Gran Variedad de Entidades

En textos más largos y complejos, el modelo puede identificar no solo nombres de personas y lugares, sino también fechas, organizaciones y otros tipos de entidades.

```python
!pip install nltk
```

```python
import nltk
from nltk import ne_chunk
from nltk.tokenize import word_tokenize
from nltk.tag import pos_tag

# Descargar recursos necesarios
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('maxent_ne_chunker')
nltk.download('words')
nltk.download('averaged_perceptron_tagger')

# Texto más complejo con varias entidades
texto = """
Apple Inc. es una empresa multinacional estadounidense que diseña, fabrica y vende dispositivos electrónicos, software y servicios. 
Fundada por Steve Jobs, Steve Wozniak y Ronald Wayne en 1976, en Cupertino, California. La compañía tiene varias sucursales en todo el mundo, incluida la tienda Apple Store en Nueva York. 
Apple también anunció que planea abrir un nuevo centro de investigación en Londres el próximo año.
"""

# Tokenizar y etiquetar las partes del habla
tokens = word_tokenize(texto)
etiquetas = pos_tag(tokens)  # Etiquetado POS usando el modelo por defecto (inglés)

# Realizar el reconocimiento de entidades nombradas
entidades = ne_chunk(etiquetas)

# Mostrar las entidades reconocidas
print("Entidades reconocidas:")
for entidad in entidades:
    if isinstance(entidad, nltk.Tree):  # Verificar si es una entidad nombrada
        print(f"Entidad: {entidad[0][0]}, Tipo: {entidad.label()}")
```

# 2. Reconocimiento de Entidades en Texto con Formato Complejo (Artículos de Noticias)
En artículos de noticias, el modelo puede reconocer no solo nombres de personas y lugares, sino también cifras importantes, eventos y otros detalles clave.

```python
texto = """
El presidente de Estados Unidos, Joe Biden, se reunió con el presidente de China, Xi Jinping, en Pekín el 4 de noviembre de 2023.
Durante la cumbre, discutieron sobre la guerra en Ucrania, el cambio climático y las relaciones comerciales entre los dos países.
El evento se celebró en el Gran Palacio del Pueblo, donde ambos líderes firmaron varios acuerdos importantes.
"""

# Tokenizar y etiquetar las partes del habla
tokens = word_tokenize(texto)
etiquetas = pos_tag(tokens)

# Realizar el reconocimiento de entidades nombradas
entidades = ne_chunk(etiquetas)

# Mostrar las entidades reconocidas
print("Entidades reconocidas:")
for entidad in entidades:
    if isinstance(entidad, nltk.Tree):  # Verificar si es una entidad nombrada
        print(f"Entidad: {entidad[0][0]}, Tipo: {entidad.label()}")
```

# 3. Reconocimiento de Entidades con Fechas y Ubicaciones Específicas
En textos con referencias a fechas y ubicaciones específicas, NLTK puede extraer información más detallada sobre eventos y ubicaciones geográficas.

```python
texto = """
El 19 de agosto de 2024, la NASA lanzará la misión Artemis II desde el Centro Espacial Kennedy en Florida. 
La misión, que tiene como objetivo llevar a los astronautas alrededor de la Luna, será un paso crucial hacia la exploración espacial más allá de la órbita terrestre.
"""

# Tokenizar y etiquetar las partes del habla
tokens = word_tokenize(texto)
etiquetas = pos_tag(tokens)

# Realizar el reconocimiento de entidades nombradas
entidades = ne_chunk(etiquetas)

# Mostrar las entidades reconocidas
print("Entidades reconocidas:")
for entidad in entidades:
    if isinstance(entidad, nltk.Tree):  # Verificar si es una entidad nombrada
        print(f"Entidad: {entidad[0][0]}, Tipo: {entidad.label()}")
```

# 4. Reconocimiento de Entidades con Enfoque en Valores Numéricos y Cantidades

Un análisis más avanzado puede involucrar la identificación de valores numéricos y cantidades, que pueden ser útiles para análisis financieros, informes técnicos, etc.

```python
texto = """
El precio de las acciones de Amazon aumentó un 5.4% en el último trimestre, alcanzando los 3,500 dólares por acción. 
Además, la compañía anunció que sus ingresos anuales superaron los 500 mil millones de dólares.
"""

# Tokenizar y etiquetar las partes del habla
tokens = word_tokenize(texto)
etiquetas = pos_tag(tokens)

# Realizar el reconocimiento de entidades nombradas
entidades = ne_chunk(etiquetas)

# Mostrar las entidades reconocidas
print("Entidades reconocidas:")
for entidad in entidades:
    if isinstance(entidad, nltk.Tree):  # Verificar si es una entidad nombrada
        print(f"Entidad: {entidad[0][0]}, Tipo: {entidad.label()}")
```

# 5. Filtrado de Entidades por Tipo

Es posible filtrar las entidades según su tipo. Por ejemplo, si solo estás interesado en personas y organizaciones, puedes extraer solo estas entidades.

```python
# Filtrar solo las entidades de tipo PERSON y ORGANIZATION
personas_y_organizaciones = [entidad[0][0] for entidad in entidades if isinstance(entidad, nltk.Tree) and entidad.label() in ['PERSON', 'ORGANIZATION']]

print("Personas y Organizaciones:", personas_y_organizaciones)
```
