```python
import requests
from bs4 import BeautifulSoup
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

# Descargar recursos de NLTK
nltk.download('punkt')
nltk.download('stopwords')

# URL de la página web a scrapear
url = 'https://es.wikipedia.org/wiki/Inteligencia_artificial'

# Hacer una solicitud GET a la página
response = requests.get(url)

# Verificar que la solicitud fue exitosa (código 200)
if response.status_code == 200:
    # Parsear el contenido HTML de la página
    soup = BeautifulSoup(response.content, 'html.parser')
    
    # Extraer el texto del contenido principal (por ejemplo, el texto dentro de las etiquetas <p>)
    paragraphs = soup.find_all('p')  # Encuentra todas las etiquetas <p> (párrafos)
    
    # Unir todo el texto de los párrafos
    text = ' '.join([para.get_text() for para in paragraphs])
    
    # Mostrar un fragmento del texto
    print(f"Texto extraído (primeros 1000 caracteres):\n{text[:1000]}...\n")
    
    # Procesar el texto con NLTK
    # Tokenizar el texto (dividirlo en palabras)
    words = word_tokenize(text)
    
    # Eliminar palabras vacías (stopwords) del texto
    stop_words = set(stopwords.words('spanish'))
    filtered_words = [word for word in words if word.lower() not in stop_words]
    
    # Mostrar las primeras 10 palabras filtradas
    print(f"Primeras 10 palabras filtradas:\n{filtered_words[:10]}")
    
else:
    print(f"Error al acceder a la página: {response.status_code}")
```
