## Primer Paso

Se importan las librerías necesarias para el procesamiento del texto y gestión de datos. Importamos json y re para trabajar con archivos JSON y re para trabajar con expresiones regulares. Después importamos pandas para poder procesar los datos utilizando DataFrames. La importación de las diferentes funciones de nltk se emplearán en el propio procesamiento de texto.

```python
# Importación de librerías necesarias
import json, re
import pandas as pd
from nltk.corpus import stopwords
from nltk.stem import SnowballStemmer
from nltk.tokenize import ToktokTokenizer
```

## Segundo paso

A continuación se abre el archivo de texto, empleando el formato json y se leen los datos del fichero. A partir de estos datos, se busca en el JSON el titular de la noticia y el cuerpo de la misma, guardándose cada una como una lista de tuplas. Posteriormente, estos datos se transforman en un dataframe con pandas.

```python
# Abrimos el archivo noticias en formato JSON 
with open('noticias.txt') as json_file:
    datos = json.load(json_file)
    
# Parseamos los datos, cargando el titular y el cuerpo de la noticia
tuplas = list(zip([noticia.get("titular") for noticia in datos],
                [noticia.get("texto") for noticia in datos]))

# Almacenamos los datos en un DataFrame de pandas
df = pd.DataFrame(tuplas, columns =['Titular', 'Noticia'])
print(df.shape)

df.head()
```

## Tercer paso

En las siguientes celdas de código, vamos a definir una función para realizar una limpieza del texto, empleando para ello funciones regulares. Esta funcción va a servir para eliminar caracteres que no nos aportan información, en este caso, aquellos no alfanuméricos que aparezcan en el texto, así como conectores de una única letra y espacios múltiples consecutivos. Por último, se convierte todo el texto en minúsculas y se devuelve el resultado.

Seguidamente, se emplea esta función sobre el dataframe para generar una nueva columna con el texto limpio, que denominaremos "Tokens".

```python
# Definimos una función para limpiar el texto
def limpiar_texto(texto):
    
    texto = re.sub(r'\W', ' ', str(texto))
    texto = re.sub(r'\s+[a-zA-Z]\s+', ' ', texto)
    texto = re.sub(r'\s+', ' ', texto, flags=re.I)
    texto = texto.lower()
    
    return texto
```

```python
#Creamos la columna tokens con los datos limpios
df["Tokens"] = df.Noticia.apply(limpiar_texto)
df.head()
```

## Cuarto Paso

En la siguiente celda de código vamos a definir la función necesaria para convertir el texto en tokens, que ha sido previamente importada de la librería nltk. Aplicaremos este método de tokenización a la columna "Tokens" de nuestro DataFrame, que contiene el texto del cuerpo de las noticias limpiado previamente. 

```python
#Completar con la explicacion
tokenizer = ToktokTokenizer()
df["Tokens"] = df.Tokens.apply(tokenizer.tokenize)
df.head()
```

## Quinto paso

En esta fase vamos a emplear las stopwords en castellano que están disponibles en la biblioteca NLTK. Estas son palabras como conectores, adverbios y preposiciones que no aportan mucha información para el procesamiento del texto. Para usar esta librería vamos a tener que descargar el paquete y establecer el conjunto de palabras para el idioma seleccionado. Después definimos una función que recorre las listas de tokens de cada noticia y los elimina si aparecen en la lista de stopwords, y aplicamos dicha función a la columna "Tokens" del DataFrame. 

```python
# Añadido para utilizar stopwords
import nltk
nltk.download('stopwords')
```

```python
#Completar con la explicacion
STOPWORDS = set(stopwords.words("spanish"))

def filtrar_stopword_digitos(tokens):
    return [token for token in tokens if token not in STOPWORDS and not token.isdigit()]

df["Tokens"] = df.Tokens.apply(filtrar_stopword_digitos)
df.head()
```

## Sexto paso

A continuación vamos a proceder con el stemming de las palabras. Este proceso supone una reducción de cada palabra a su raíz semántica, también denominado lexema. Para ello vamos a el método SnowballStemmer de la librería NLTK, ajustado para el idioma castellano. Creamos una función que recorra una lista de tokens y extraiga su lexema, y aplicamos dicha función a columna con las listas de tokens de cada noticia. Por último, mostramos los primeros 10 resultados de la primera noticia, para ver el efecto del stemming.

```python
#Completar con la explicación
stemmer = SnowballStemmer("spanish")
def stem_palabras(tokens):
    """
    Reduce cada palabra de una lista dada a su raíz.
    """
    return [stemmer.stem(token) for token in tokens]

df["Tokens"] = df.Tokens.apply(stem_palabras)
df.head()
```

```python
#Completar con la explicación
print(df.Tokens[0][0:10])
```
