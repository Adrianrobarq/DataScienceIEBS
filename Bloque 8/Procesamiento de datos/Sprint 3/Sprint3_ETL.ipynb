En este sprint, vamos a emplear la API de <a href=https://developer.edamam.com/edamam-docs-recipe-api>Recipe Search</a> para buscar recetas de Camembert.

```python
# Importamos librerías necesarias
import requests
import urllib.parse
```

```python
# Introducimos parámetros de la API

endpoint = 'https://api.edamam.com/api/recipes/v2'

id = '10955bb8'
key = '2a2e54195f32378a6d30958ebb4597ca'

params = {'type': 'public',
        'q': 'Camembert',
        'calories': '500+',
}
```

```python
# Construimos la URL de la API
url = f'{endpoint}?{urllib.parse.urlencode(params)}'

autothorize = url+f'&app_id={id}&app_key={key}'
```

```python
print(autothorize)
```

```python
# Añadimos el header para la autorización
headers = {
    'accept': 'application/json',
    'Accept-Language': 'en',
    'Edamam-Account-User': 'KadrenExal'
}

# Realizamos la petición
response = requests.get(autothorize, headers=headers)

# Verificamos el estado de la respuesta
if response.status_code == 200:
    data = response.json()
    print("Petición exitosa")
    print(f"Número de recetas encontradas: {len(data.get('hits', []))}")
else:
    print(f"Error: {response.status_code}")
    print(response.text)
```

```python
# Extraemos las recetas
recipes = data.get('hits', [])

# Hacemos un diccionario con el nombre de la receta y su coste calórico
recipes_info = {recipe['recipe']['label']: recipe['recipe']['calories'] for recipe in recipes}
```

```python
# Importamos pandas para crear un DataFrame
import pandas as pd

# Creamos un DataFrame con la información de las recetas
df_recipes = pd.DataFrame([[k, v] for k, v in recipes_info.items()], columns=['Receta', 'Calorias'])

# Ordenamos el DataFrame por calorías de forma descendente
df_recipes_sorted = df_recipes.sort_values(by='Calorias', ascending=False).reset_index(drop=True)

# Mostramos el DataFrame ordenado
df_recipes_sorted.head()
```

El primer resultado de la tabla es, por tanto, la receta con más contenido calórico total.
