Implementa un script en Python para extraer mediante Web Scraping el valor total de la inversión, calculado como:

VALOR TOTAL = ∑ fondo x participaciones
PARTICIPACIONES:
* BlackRock Global Funds - Euro Short Duration Bond Fund A2 EUR: https://es.investing.com/funds/euro-short-duration-bond-fund-a2-eu-scoreboard
* iShares Europe Equity Index Fund (LU) A2 EUR: https://es.investing.com/funds/ishares-europe-equity-index-a2
* iShares Euro Government Bond Index Fund (LU) A2 EUR:  https://es.investing.com/funds/blackrock-euro-government-bond-indn-historical-data

```python
# Importamos las librerías necesarias

import urllib.request as urlreq
from bs4 import BeautifulSoup
import pandas as pd
```

```python
# Almacenamos las URLs en un diccionario
urls = {
    'Euro Short Bond':  'https://es.investing.com/funds/euro-short-duration-bond-fund-a2-eu-scoreboard',
    'Europe Equity Index': 'https://es.investing.com/funds/ishares-europe-equity-index-a2',
    'Euro Government Bond': 'https://es.investing.com/funds/blackrock-euro-government-bond-indn-historical-data'
}
```

```python
# Extraemos el contenido de las páginas web
def extract_data(url, tag, element_id):
    
    # Usamos agente para evitar bloqueo del servidor
    request = urlreq.Request(url, headers={
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36'
    })
    
    html = urlreq.urlopen(request)
    soup = BeautifulSoup(html, 'html.parser')
    
    indice = soup.find(tag, id= element_id)
    
    # Convertimos el contenido a float
    valor = indice.text.strip().replace(',', '.')
    
    return float(valor)
```

```python
tag = 'span'
element_id = 'last_last'
indices = {name: extract_data(url, tag, element_id) for name, url in urls.items()}
```

```python
indices
```

```python
# Definimos las participaciones que tenemos en cada fondo
participaciones = {
    'Euro Short Bond': 22.45,
    'Europe Equity Index': 8.34,
    'Euro Government Bond': 13.22
}
```

```python
# Convertimos en tabla de Pandas
df = pd.DataFrame(list(participaciones.items()), columns=['Fondo', 'Participaciones'])
df['Valor Actual'] = df['Fondo'].map(indices)
df['Valor Total'] = df['Valor Actual'] * df['Participaciones']
df.set_index('Fondo', inplace=True)
df
```

```python
# Calculamos el valor total de la cartera de inversiones
valor_total_portafolio = df['Valor Total'].sum()
print(f'Valor total del portafolio: {valor_total_portafolio:.2f} €')
```
