## ¿Qué es un decorador?

Un decorador es una función que toma otra función como argumento y extiende su comportamiento sin modificarla explícitamente. Los decoradores son una forma elegante de aplicar el patrón de diseño "wrapper" o "decorator".

**Sintaxis básica:**
```python
@decorador
def funcion():
    pass
```

Es equivalente a:
```python
def funcion():
    pass
funcion = decorador(funcion)
```

## Ejercicio 1: Tu Primer Decorador Simple

**Objetivo:** Crear un decorador básico que imprima un mensaje antes y después de ejecutar una función.

**Instrucciones:**
1. Completa la función `mi_decorador` que debe:
   - Imprimir "Antes de ejecutar la función"
   - Ejecutar la función original
   - Imprimir "Después de ejecutar la función"
2. Aplica el decorador a la función `saludar()`

### Solución Ejercicio 1

Aquí está la solución completa:

````python
def mi_decorador(funcion):
    def envoltura():
        print("Antes de ejecutar la función")
        funcion()
        print("Después de ejecutar la función")
    return envoltura

@mi_decorador
def saludar():
    print("¡Hola mundo!")

# Prueba la función
saludar()
````

## Ejercicio 2: Decorador con Argumentos

**Objetivo:** Crear un decorador que funcione con funciones que reciben parámetros.

**El Problema:** El decorador anterior solo funciona con funciones sin parámetros. Necesitamos usar `*args` y `**kwargs`.

**Instrucciones:**
1. Modifica el decorador para que acepte funciones con cualquier número de argumentos
2. Aplica el decorador a una función que reciba parámetros

### Solución Ejercicio 2

````python
def decorador_flexible(funcion):
    def envoltura(*args, **kwargs):  # Acepta cualquier argumento
        print(f"Ejecutando: {funcion.__name__}")
        resultado = funcion(*args, **kwargs)  # Pasa los argumentos a la función original
        return resultado
    return envoltura

@decorador_flexible
def sumar(a, b):
    resultado = a + b
    print(f"Resultado: {resultado}")
    return resultado

@decorador_flexible
def saludar_persona(nombre, edad=None):
    if edad:
        print(f"Hola {nombre}, tienes {edad} años")
    else:
        print(f"Hola {nombre}")

# Pruebas
sumar(5, 3)
print()
saludar_persona("Ana")
print()
saludar_persona("Carlos", edad=30)
````

## Ejercicio 3: Decorador de Tiempo de Ejecución

**Objetivo:** Crear un decorador práctico que mida cuánto tiempo tarda en ejecutarse una función.

**Instrucciones:**
1. Usa el módulo `time` para medir el tiempo de ejecución
2. El decorador debe mostrar cuántos segundos tardó la función
3. Aplícalo a funciones que tarden diferentes tiempos

### Solución Ejercicio 3

````python
import time

def medir_tiempo(funcion):
    def envoltura(*args, **kwargs):
        inicio = time.time()  # Tiempo inicial
        resultado = funcion(*args, **kwargs)
        fin = time.time()  # Tiempo final
        tiempo_transcurrido = fin - inicio
        print(f"⏱️  '{funcion.__name__}' tardó {tiempo_transcurrido:.4f} segundos")
        return resultado
    return envoltura

@medir_tiempo
def funcion_rapida():
    suma = sum(range(100000))
    return suma

@medir_tiempo
def funcion_lenta():
    time.sleep(2)
    return "¡Terminé!"

# Prueba ambas funciones
resultado1 = funcion_rapida()
print(f"Resultado: {resultado1}\n")

resultado2 = funcion_lenta()
print(f"Resultado: {resultado2}")
````

## Ejercicio 4: Múltiples Decoradores

**Concepto Importante:** Puedes aplicar varios decoradores a la misma función. Se ejecutan de **abajo hacia arriba**.

**Ejemplo:**
```python
@decorador1
@decorador2
@decorador3
def funcion():
    pass
```

Es equivalente a: `decorador1(decorador2(decorador3(funcion)))`

**Instrucciones:**
1. Crea un decorador `convertir_mayusculas` que convierta el resultado a mayúsculas
2. Crea un decorador `agregar_signos` que agregue "¡¡¡" al inicio y "!!!" al final
3. Aplica ambos decoradores a una función

### Solución Ejercicio 4

````python
def convertir_mayusculas(funcion):
    def envoltura(*args, **kwargs):
        resultado = funcion(*args, **kwargs)
        return resultado.upper()
    return envoltura

def agregar_signos(funcion):
    def envoltura(*args, **kwargs):
        resultado = funcion(*args, **kwargs)
        return f"¡¡¡{resultado}!!!"
    return envoltura

# Orden 1: signos primero, luego mayúsculas
@convertir_mayusculas
@agregar_signos
def mensaje1():
    return "esto es importante"

# Orden 2: mayúsculas primero, luego signos
@agregar_signos
@convertir_mayusculas
def mensaje2():
    return "esto es importante"

print("Orden 1 (signos → mayúsculas):", mensaje1())
print("Orden 2 (mayúsculas → signos):", mensaje2())
````

## Ejercicio 5: Decoradores con Parámetros (Nivel Avanzado)

**Concepto:** A veces queremos que el decorador mismo reciba parámetros. Para esto necesitamos **tres niveles de funciones**.

**Estructura:**
```python
def decorador_con_parametros(parametro):
    def decorador_real(funcion):
        def envoltura(*args, **kwargs):
            # Usar 'parametro' aquí
            return funcion(*args, **kwargs)
        return envoltura
    return decorador_real
```

**Instrucciones:**
1. Crea un decorador `repetir` que reciba un número y ejecute la función ese número de veces
2. Aplícalo a una función con diferentes valores

### Solución Ejercicio 5

````python
def repetir(veces):
    def decorador_real(funcion):
        def envoltura(*args, **kwargs):
            for i in range(veces):
                resultado = funcion(*args, **kwargs)
            return resultado  # Retorna el resultado de la última ejecución
        return envoltura
    return decorador_real

@repetir(3)
def saludar(nombre):
    print(f"¡Hola {nombre}!")

@repetir(5)
def imprimir_asterisco():
    print("*", end="")

# Prueba
saludar("María")
print()
imprimir_asterisco()
print()
````

## Ejercicio 6: `functools.wraps` - Preservar Metadatos

**Problema:** Los decoradores cambian el nombre y la documentación de la función original.

**Solución:** Usar `@functools.wraps(funcion)` en la función envoltura.

**Instrucciones:**
1. Observa el problema sin usar `functools.wraps`
2. Compáralo con la versión correcta que sí lo usa

SOLUCIÓN EJERCICIO 6

````python
from functools import wraps

# Decorador SIN functools.wraps (problemático)
def mi_decorador_malo(funcion):
    def envoltura(*args, **kwargs):
        """Documentación de la envoltura"""
        return funcion(*args, **kwargs)
    return envoltura

# Decorador CON functools.wraps (correcto)
def mi_decorador_bueno(funcion):
    @wraps(funcion)  # ¡Esto preserva los metadatos!
    def envoltura(*args, **kwargs):
        """Documentación de la envoltura"""
        return funcion(*args, **kwargs)
    return envoltura

@mi_decorador_malo
def funcion_mala(x, y):
    """Esta función suma dos números"""
    return x + y

@mi_decorador_bueno
def funcion_buena(x, y):
    """Esta función suma dos números"""
    return x + y

# Compara los resultados
print("SIN functools.wraps:")
print(f"  Nombre: {funcion_mala.__name__}")
print(f"  Documentación: {funcion_mala.__doc__}")
print()
print("CON functools.wraps:")
print(f"  Nombre: {funcion_buena.__name__}")
print(f"  Documentación: {funcion_buena.__doc__}")
````

## Ejercicio 7: Decorador de Registro (Logging)

**Objetivo:** Crear un decorador práctico que registre las llamadas a funciones con sus argumentos y resultados.

**Instrucciones:**
1. El decorador debe mostrar:
   - El nombre de la función
   - Los argumentos con los que se llamó
   - El resultado que devolvió
2. Úsalo con diferentes funciones para ver cómo funciona

### Solución Ejercicio 7

````python
from functools import wraps

def registrar(funcion):
    @wraps(funcion)
    def envoltura(*args, **kwargs):
        # Registrar la llamada
        args_repr = [repr(a) for a in args]
        kwargs_repr = [f"{k}={v!r}" for k, v in kwargs.items()]
        signature = ", ".join(args_repr + kwargs_repr)
        
        print(f"📝 Llamando a {funcion.__name__}({signature})")
        
        # Ejecutar la función
        resultado = funcion(*args, **kwargs)
        
        # Registrar el resultado
        print(f"✅ {funcion.__name__} retornó {resultado!r}")
        
        return resultado
    return envoltura

@registrar
def calcular_area_rectangulo(base, altura):
    return base * altura

@registrar
def dividir(a, b):
    return a / b

# Prueba
resultado1 = calcular_area_rectangulo(5, 3)
print()
resultado2 = dividir(10, 2)
````

## Ejercicio 8: Decorador de Caché (Memoización)

**Concepto:** La memoización guarda los resultados de funciones costosas para evitar recalcularlos.

**Caso de Uso:** Funciones recursivas como Fibonacci son extremadamente lentas sin caché.

**Instrucciones:**
1. Crea un decorador que guarde los resultados en un diccionario
2. Si la función se llama con los mismos argumentos, devuelve el resultado guardado
3. Compara el rendimiento con y sin caché en la función de Fibonacci

### Solución Ejercicio 8

````python
import time
from functools import wraps

def cache(funcion):
    cache_dict = {}
    
    @wraps(funcion)
    def envoltura(*args):
        if args in cache_dict:
            print(f"💾 Usando caché para {args}")
            return cache_dict[args]
        
        print(f"🔄 Calculando para {args}")
        resultado = funcion(*args)
        cache_dict[args] = resultado
        return resultado
    
    return envoltura

# Fibonacci SIN caché
def fibonacci_sin_cache(n):
    if n < 2:
        return n
    return fibonacci_sin_cache(n-1) + fibonacci_sin_cache(n-2)

# Fibonacci CON caché
@cache
def fibonacci_con_cache(n):
    if n < 2:
        return n
    return fibonacci_con_cache(n-1) + fibonacci_con_cache(n-2)

# Prueba la diferencia (usa números más pequeños para no esperar demasiado)
print("Calculando fibonacci(10) SIN caché...")
inicio = time.time()
resultado1 = fibonacci_sin_cache(10)
tiempo1 = time.time() - inicio
print(f"Resultado: {resultado1}, Tiempo: {tiempo1:.6f}s\n")

print("Calculando fibonacci(10) CON caché...")
inicio = time.time()
resultado2 = fibonacci_con_cache(10)
tiempo2 = time.time() - inicio
print(f"Resultado: {resultado2}, Tiempo: {tiempo2:.6f}s\n")

if tiempo2 > 0:
    print(f"🚀 ¡El caché es {tiempo1/tiempo2:.0f}x más rápido!")
````

## Bonus: `functools.lru_cache`

Python incluye un decorador de caché integrado y optimizado: `@lru_cache` (Least Recently Used cache).

**Ventajas:**
- Ya está implementado y optimizado
- Soporta un límite de tamaño (`maxsize`)
- Funciona con argumentos keyword

**Uso:**

````python
from functools import lru_cache
import time

@lru_cache(maxsize=128)  # Guarda hasta 128 resultados
def fibonacci_lru(n):
    if n < 2:
        return n
    return fibonacci_lru(n-1) + fibonacci_lru(n-2)

# Prueba con un número grande
print("Calculando fibonacci(100) con lru_cache...")
inicio = time.time()
resultado = fibonacci_lru(100)
tiempo = time.time() - inicio

print(f"Resultado: {resultado}")
print(f"Tiempo: {tiempo:.6f}s")

# Ver estadísticas del caché
print(f"\nEstadísticas del caché:")
print(fibonacci_lru.cache_info())
````
