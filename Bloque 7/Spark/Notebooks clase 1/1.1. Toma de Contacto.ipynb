# 2. Creación de un RDD desde una lista

```python
# Crear un RDD desde una lista de Python
data = [1, 2, 3, 4, 5]
rdd = spark.sparkContext.parallelize(data)

# Mostrar el RDD
print(f"RDD inicial: {rdd.collect()}")
```

# 3. Operaciones básicas de RDD: map, filter y reduce

```python
# Aplicar la operación map para multiplicar cada elemento por sí mismo
squared_rdd = rdd.map(lambda x: x * x)
print(f"RDD con elementos al cuadrado: {squared_rdd.collect()}")

# Filtrar los elementos mayores a 2
filtered_rdd = rdd.filter(lambda x: x > 2)
print(f"RDD filtrado (elementos mayores a 2): {filtered_rdd.collect()}")

# Sumar todos los elementos del RDD
sum_rdd = rdd.reduce(lambda a, b: a + b)
print(f"Suma de los elementos del RDD: {sum_rdd}")
```

# 4. Creación de un RDD a partir de datos falsos generados

```python
# Crear un RDD de datos falsos (nombres y edades)
fake_data = [("Juan", 28), ("Ana", 23), ("Luis", 35), ("Marta", 30), ("Pedro", 40)]
rdd_fake = spark.sparkContext.parallelize(fake_data)

# Mostrar el RDD generado
print(f"Datos falsos: {rdd_fake.collect()}")
```

# 5. Ejemplo práctico: Filtrar y transformar datos

```python
# Filtrar las personas mayores de 30 años
rdd_mayores_30 = rdd_fake.filter(lambda x: x[1] > 30)
print(f"Personas mayores de 30 años: {rdd_mayores_30.collect()}")

# Aplicar una transformación para crear mensajes de saludo
rdd_saludos = rdd_fake.map(lambda x: f"Hola, {x[0]}! Tienes {x[1]} años.")
print(f"Mensajes de saludo: {rdd_saludos.collect()}")
```

# 6. Uso de RDD para cálculos sencillos

```python
# Calcular el promedio de las edades en el RDD
edades = rdd_fake.map(lambda x: x[1])
total_edad = edades.reduce(lambda a, b: a + b)
promedio_edad = total_edad / edades.count()

print(f"Promedio de edad: {promedio_edad}")
```

# 7. Acciones avanzadas: count, take y foreach

```python
# Contar el número de elementos en el RDD
print(f"Número de personas: {rdd_fake.count()}")

# Tomar los primeros 3 elementos del RDD
print(f"Primeras 3 personas: {rdd_fake.take(3)}")

# Aplicar foreach para imprimir cada elemento
rdd_fake.foreach(lambda x: print(f"Nombre: {x[0]}, Edad: {x[1]}"))
```
